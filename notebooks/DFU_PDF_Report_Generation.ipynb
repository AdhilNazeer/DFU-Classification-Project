```python

from pathlib import Path

# Main project folder
PROJECT_DIR = Path(
    r"C:\Users\adhil\OneDrive\Desktop\dfu\DFU_Classification_Project"
)

# Saved trained model
MODEL_PATH = PROJECT_DIR / "models" / "dfu_grade_model_best.keras"

# Folder for evaluation outputs and PDF
OUTPUT_DIR = PROJECT_DIR / "outputs" / "evaluation"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# Check that the model exists
print("Project folder:", PROJECT_DIR)
print("Model exists:", MODEL_PATH.exists())
print("Output folder:", OUTPUT_DIR)
```

```python

import tensorflow as tf
from tensorflow import keras

# Dataset folders
DATA_DIR = PROJECT_DIR / "data" / "dfu 4 class"
TEST_DIR = DATA_DIR / "test"

# Image settings used by your model
IMG_SIZE = (300, 300)
BATCH_SIZE = 16

# Grade names in the expected class order
class_names = ["Grade 1", "Grade 2", "Grade 3", "Grade 4"]

# Load the saved trained model
best_model = keras.models.load_model(MODEL_PATH)

print("Model loaded successfully!")

# Create the test dataset
test_ds = tf.keras.utils.image_dataset_from_directory(
    TEST_DIR,
    labels="inferred",
    label_mode="int",
    class_names=class_names,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False
)

print("Test dataset created successfully!")
print("Class order:", test_ds.class_names)
print("Number of test images:", len(test_ds.file_paths))
```
